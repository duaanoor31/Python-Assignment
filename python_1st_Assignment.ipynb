```python
#Qno1 Take a number from the user and check whether it is Even or Odd
number = int(input ('Enter a number:'))
if number % 2 == 0:
  print('Even number')
else:
  print("Odd number")
```

```python
#Qno2 Take two numbers and an operator (+,-,*,/) from the user.Perform the selected operation and display the result
num1 = int(input('Enter first number:'))
num2 = int(input('Enter second number'))
operator = input("Enter operator(+,-,*,/):")
if operator =='+':
           print(num1 + num2)
elif operator =='-':
           print(num1 - num2)
elif operator =='*':
           print(num1 * num2)
elif operator =='/':
           print(num1 / num2)
else:
           print('invalid operator')
```

```python
#Qno3 Take marks of 3 subjects from the user.Calculate total marks and percentage.Display "Pass" if percentage is 50 or above, otherwise display "Fail"
sub1 = int(input("Enter marks of subject 1 :"))
sub2 = int(input("Enter marks of subject 2:"))
sub3 = int(input("Enter marks of subject 3:"))
total = sub1 + sub2 + sub3
percentage = (total / 300) * 100
print(total)
print(percentage)
if percentage >= 50:
  print("Pass")
else:
  print("Fail")

```

```python
#Qno4 create a list of 5 fruits.Ask the user to enter a fruit and check whether it exists in the list.
fruits = ["Apple","Banana","Mango","orange","Grapes"]
fruit = input("Enter a fruit name:")
if fruit in fruits:
  print("Fruit exists in the list")
else:
  print("Fruit is not exists in the list")
```
