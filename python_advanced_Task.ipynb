```python
#Task 11
#Personal Expense Tracker:
expenses = []
total = int(input("Enter total number of expenses:"))
for i in range (total):
  print(f"expense" , i + 1)
  name = input("Enter expense name:")
  category = input("Enter category = (Food/Transport/Shopping/Bills/Other:)")
  amount = float(input("Enter amount:"))
  expense = {
      "Name" : name,
      "Category" : category,
      "Amount" : amount,
      }
  expenses.append(expense)
def expense_report():
  total_spending = 0
  category_total = {}
  if not expenses:
      print("No expenses to report.")
      return
  highest = expenses[0]
  for expense in expenses:
    total_spending += expense["Amount"]
    cat = expense["Category"]
    if cat in category_total:
      category_total[cat] += expense["Amount"]
    else:
      category_total[cat] = expense["Amount"]
    if expense["Amount"] > highest["Amount"]:
      highest = expense

  print("===============FINAL REPORT============")
  print("Total Spending = ", total_spending)
  print(f"category wise spending:")
  for cat in category_total:
    print(cat, "=", category_total[cat])
  print(f"Highest Expense:")
  print(highest["Name"], "=", highest["Amount"])

  budget = float(input("Enter  your Budget:"))
  if total_spending > budget:
    print("Warning! Budget Exceeded:")
  else:
    print("Good! You are within your budget.")
expense_report()
```

```python
#Task 12
# Store Inventory Manager:
inventory = {}
def add_product():
    name = input("Enter Product Name: ")

    if name in inventory:
        print("Product already exists!")
    else:
        price = float(input("Enter Price: "))
        quantity = int(input("Enter Quantity: "))

        inventory[name] = {
            "price": price,
            "quantity": quantity
        }

        print("Product Added Successfully!")


def sell_product():
    name = input("Enter Product Name: ")

    if name in inventory:
        qty = int(input("Enter Quantity to Sell: "))

        if inventory[name]["quantity"] >= qty:
            inventory[name]["quantity"] -= qty
            print("Product Sold Successfully!")
        else:
            print("Warning! Insufficient Stock.")
    else:
        print("Product Not Found!")
def restock_product():
    name = input("Enter Product Name: ")

    if name in inventory:
        qty = int(input("Enter Quantity to Add: "))
        inventory[name]["quantity"] += qty
        print("Stock Updated Successfully!")
    else:
        print("Product Not Found!")

def search_product():
    name = input("Enter Product Name: ")

    if name in inventory:
        print("\nProduct Found")
        print("Price:", inventory[name]["price"])
        print("Quantity:", inventory[name]["quantity"])
    else:
        print("Product Not Found!")

def display_inventory():

    if len(inventory) == 0:
        print("Inventory is Empty!")
        return

    total_value = 0

    print("\n------ Inventory ------")

    for name, details in inventory.items():

        price = details["price"]
        quantity = details["quantity"]
        value = price * quantity

        total_value += value

        print("\nProduct:", name)
        print("Price:", price)
        print("Quantity:", quantity)
        print("Value:", value)

        if quantity < 5:
            print("Low Stock!")

    print("\nTotal Inventory Value =", total_value)
while True:

    print("\n===== STORE INVENTORY MANAGER =====")
    print("1. Add Product")
    print("2. Sell Product")
    print("3. Restock Product")
    print("4. Search Product")
    print("5. Display Inventory")
    print("6. Exit")

    choice = input("Enter Your Choice: ")

    if choice == "1":
        add_product()

    elif choice == "2":
        sell_product()

    elif choice == "3":
        restock_product()

    elif choice == "4":
        search_product()

    elif choice == "5":
        display_inventory()

    elif choice == "6":
        print("Program Closed.")
        break

    else:
        print("Invalid Choice!")
```

```python
#Task 13
#Library Book Managment System:
books = []
def add_book():
  book_id = input("Enter book ID:")
  for book in books:
    if book["id"] == book_id:
      print("book ID already exists!")
      return
  title = input("Enter Book Title:")
  author = input("Enter Author Name:")
  book = {
      "id" : book_id,
      "title" : title,
      "author" : author,
      "status" : "Available"
  }
  books.append(book)
  print("Book Added Successfully!")

def display_books ():
  if len(books) == 0:
    print("No Books Available!")
    return
  print("=================Library Books===============")
  for book in books:
    print("===========================")
    print("Book ID:" , book["id"])
    print("Title:" , book["title"])
    print("Author:" , book["author"])
    print("Status" , book["status"])

def search_book():
  title = input("Enter Book Title:")
  for book in books:
    if book["title"].lower() == title.lower():
      print(f"Book Found!")
      print("Book ID:" , book["id"])
      print("Title:" , book["title"])
      print("Author:" , book["author"])
      print("Status" , book["status"])
      return
      print("Book Not Found!")
def borrow_book():
  book_id = input("Enter Book ID:")
  for book in books:
    if book["id"] == book_id:
            if book["status"] == "Borrowed":
                book["status"] = "Available"
                print("Book Returned Successfully!")
            else:
                print("Book is Already Available!")
            return

    print("Book Not Found!")
def return_book():
    book_id = input("Enter Book ID: ")

    for book in books:
        if book["id"] == book_id:
            if book["status"] == "Borrowed":
                book["status"] = "Available"
                print("Book Returned Successfully!")
            else:
                print("Book is Already Available!")
            return

    print("Book Not Found!")

while True:

    print("\n===== Library Book Management System =====")
    print("1. Add Book")
    print("2. Display Books")
    print("3. Search Book")
    print("4. Borrow Book")
    print("5. Return Book")
    print("6. Exit")

    choice = input("Enter Your Choice: ")

    if choice == "1":
        add_book()

    elif choice == "2":
        display_books()

    elif choice == "3":
        search_book()

    elif choice == "4":
        borrow_book()

    elif choice == "5":
        return_book()

    elif choice == "6":
        print("Program Closed.")
        break

    else:
        print("Invalid Choice!")


```

```python
#TASK 14 Restaurant ordering system:
menu= {"Burger":500,"Pizza":1200,"Pasta":800,"Drink":200}
orders = []
print("=========Restaurant Menu===========")
for key, value in menu.items():
  print(f"{key}: {value}")

while True:
  item_name = input("Enter item name:").title()
  if item_name in menu:
    quantity = int(input(f"Enter the quantity of {item_name}:"))
    price = menu[item_name]
    sub_total = price * quantity
    print(f"Total amount:{sub_total}")
    orders.append({
        "item_name": item_name,
        "quantity" : quantity,
        "price" : price,
        "sub_total" : sub_total,
    })
  else:
    print(f"This {item_name} is not available!")

  choice = input("Do you want to add more items (yes/no)? ")
  if choice.lower() != "yes":
    break

total = 0
for order in orders:
  total += order["sub_total"]

tax = total * 13/100
if total >= 3000:
  discount = total * 10/100
else:
  discount = 0
grand_total = total + tax - discount
def receipt():
  print("===========Receipt============")
  for order in orders:
    print(f"{order["item_name"]} * {order["quantity"]} = {order["sub_total"]}")
    print(f"Total: {total}")
    print(f"Tax: {tax}")
    print(f"Discount: {discount}")
    print(f"Grand Total: {grand_total}")

receipt()
```

```python
# Task 15 clinic Appointment Manager:
patients = []
def add_patient():
  patient_id = int(input("Enter your id:"))
  for patient in patients:
    if patient["id"] == patient_id:
      print("patient already exists")
      return
  patient_name = input("Enter patient name:")
  patient_age = int(input("Enter patient age:"))
  problem = input("Enter patient problem:")
  time = input("Enter Appointment time:")
  emergency = input("is emergency patient? yes/no:").lower()
  patients.append({
      "id": patient_id,
      "name": patient_name,
      "age": patient_age,
      "problem": problem,
      "time": time,
      "emergency": emergency
  })
def search_patient():
  patient_id = int(input("Enter patient id:"))
  for patient in patients:
    if patient["id"] == patient_id:
      print(f"patient found")
      print(f"patient name: {patient["name"]}")
      print(f"patient age: {patient["age"]}")
      print(f"patient problem: {patient["problem"]}")
      print(f"pateint time: {patient["time"]}")
      print(f"patient emergency: {patient["emergency"]}")
      return
  else:
    print("patient not found")
def display_appointment():
  print("=============Emergency patient===========")
  for patient in patients:
    if patient["emergency"] == "yes":
      print(f"patient name: {patient["name"]}")
      print(f"patient age: {patient["age"]}")
      print(f"patient problem: {patient["problem"]}")
      print(f"patient time: {patient["time"]}")
  print("==============normal patient===============")
  for patient in patients:
    if patient["emergency"] == "no":
      print(f"patient name: {patient["name"]}")
      print(f"patient age: {patient["age"]}")
      print(f"patient problem: {patient["problem"]}")
      print(f"patient time: {patient["time"]}")
def delete_patient():
  patient_id = int(input("Enter patient id:"))
  for i, patient in enumerate(patients):
    if patient["id"] == patient_id:
      patients.pop(i)
      print("patient deleted")
      return
  print("patient not found")
def count_patient():
  emergency_count = 0
  normal_count = 0
  for patient in patients:
    if patient["emergency"] == "yes":
      emergency_count += 1
    else:
      normal_count += 1
  print(f"Total emergency patients: {emergency_count}")
  print(f"Total normal patients: {normal_count}")
while True:
  print("1. Add patient")
  print("2. Search patient")
  print("3. Display appointment")
  print("4. Delete patient")
  print("5. Count patient")
  print("6. Exit")
  choice = int(input("Enter your choice:"))
  if choice == 1:
    add_patient()
  elif choice == 2:
    search_patient()
  elif choice == 3:
    display_appointment()
  elif choice == 4:
    delete_patient()
  elif choice == 5:
    count_patient()
  elif choice == 6:
    break
  else:
    print("invalid choice")
```
