```python
#Task 4
#weekly steps checker
total_steps = 0
for i in range(7):
  steps = int(input("Enter the steps:"))
  total_steps += steps
average = steps // 7
print(f"total steps:{total_steps}")
print(f"Daily average: {average}")
if(average >= 5000):
  print("status: Active week")
else:
  print("status: needs more activity")


```

```python
#Task 1
#cinema ticket price
Name = input("Enter your name ")
age = int(input("Enter your age "))
tick_price = 0
if age <= 5 and age <= 100:
 tick_price = 0
 print("it's  free")
elif age >= 5 and age <= 12:
  tick_price = 300
elif age >= 13 and age <= 59:
 tick_price = 700
elif age >= 60 and age <= 100:
  tick_price = 400
else:
  print("Invalid age ")

print(f"customer name : {Name}")
print(f"age : {age}")
print(f"Ticket price: {tick_price}")
```

```python
#Task 2
#Temperature converter
celsius = float(input("Enter temperature in celsius:"))
fahrenheit = (celsius * 9/5) + 32
print("Temperature in fahrenheit:" , fahrenheit)
if celsius < 10:
  print("weather: very cold")
elif celsius >= 10 and celsius <= 19:
  print("weather: cold")
elif celsius >= 20 and celsius <= 29:
  print("weather: Normal")
else:
  print("weather: Hot")
```
